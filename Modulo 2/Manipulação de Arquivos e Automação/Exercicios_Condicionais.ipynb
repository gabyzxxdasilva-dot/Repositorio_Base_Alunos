Exercício 4 - Calculadora de Desconto. 💵 🤑

Uma loja dá 10% de desconto para compras acima de R$ 100,00. Faça um programa que:

1º - Pergunta o valor da compra.

2º - Calcula o valor final com desconto (se aplicável).

3º - Mostra o valor a pagar.

```python

valor = float(input("Informe o valor da compra."))

if valor >100.00:
    desconto = valor * 0.10
    print(f' Voce fez {valor} em compras, teve um desconto de {desconto}, total a pagar {valor - desconto}')
else:
    print(" voce não teve desconto")

```

Exercício 5 - Classificação de Notas 🧑‍🎓

Pergunte a nota de um aluno e dependendo da nota que ele informar, o classifique de acordo com as regras abaixo:

Classifique a nota do aluno:

- 9 até 10: "Excelente"

- 7 até 8.9: "Bom"

- 5 até 6.9: "Regular"

- 0 até 4.9: "Reprovado"

mostre para o aluno a classificação dele através de um print, exemplo print("Excelente").

```python
nota = float(input("Qual a sua nota ? :"))

if nota >=9 and nota <= 10:
    print("Excelente!!")

elif nota >=7 and nota <=8.9:
    print("bom")

elif nota >=5 and nota <=6.9:
    print("regular")

elif nota >=0 and nota <=4.9:
    print("reprovado")
 
else:
     print("nota ivalida")


```

Exercício 8 - Dia da Semana
Peça um número de 1 a 7 e mostre o dia da semana correspondente: 📅

- 1: Domingo

- 2: Segunda

- ... até 7: Sábado

```python
numero = int(input("Digite um número de 1 até 7"))

if numero ==1 :
    print("Domingo")
elif numero ==2 :
    print("Segunda")
elif numero ==3:
    print("Terça")
elif numero ==4 :
    print("Quarta")
elif numero ==5 :
    print("Quinta")
elif numero ==6 :
    print("Sexta")
elif numero ==7 : 
    print("Sabado")
else :
    print("Incorreto")
```

Exercício 9 - Calculadora Simples 📏📐

- Faça uma calculadora que:

- Pede dois números

- Pede a operação (+, -, *, /)

- Dependendo da operação mostre o resultado.



```python
primeiro_numero = float(input(" Digite o seu prinmeiro numero"))
segundo_numero = float(input(" Digite o seu segundo numero"))
operacao = input("digite  a operação :(+,-,*,/)")
if operacao =="+" :
    print(f' A Soma de {primeiro_numero} + {segundo_numero} e igual a {primeiro_numero+segundo_numero}')
elif operacao =="-" :
     print(f' A Subitração de {primeiro_numero} - {segundo_numero} e igual a {primeiro_numero-segundo_numero}')
elif operacao =="*" :
     print(f' A Mutiplicação de {primeiro_numero} * {segundo_numero} e igual a {primeiro_numero*segundo_numero}')
elif operacao =="/" :
     print(f' A Divisão de {primeiro_numero} / {segundo_numero} e igual a {primeiro_numero/segundo_numero}')
     
```

Exercício 10 - Sistema de Multas 🚙🚘🧾


- Velocidade máxima: 80 km/h 🚫

- Até 80: "Sem multa"

- 81-100: "Multa leve"

- Acima de 100: "Multa grave"

```python
velocidade = int(input("Digite a sua velocidade"))
if velocidade <=80 :
    print("sem multa")

elif velocidade >=81 and velocidade<=100:
    print("multa leve ")

else:
    print("multa grave")
```
